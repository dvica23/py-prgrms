```python
n = int(input("Enter a number: "))
rev = 0
while n > 0:
    rem = n % 10
    rev = rev * 10 + rem
    n = n // 10
print("Reversed number =", rev)
```

```python
n = int(input("Enter a number: "))
sum = 0
while n > 0:
    rem = n % 10
    sum = sum + rem
    n = n // 10
print("Sum of digits =", sum)
```

```python
n = int(input("Enter a number: "))
largest = 0
while n > 0:
    digit = n % 10
    if digit > largest:
        largest = digit
    n = n // 10
print("largest digit =",largest)
```

```python
n = int(input("Enter a number: "))
num = n
rev = 0
while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10
if num == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
str = input("Enter a string: ")
count = 0
vow = 0
cons = 0
for char in string:
    count = count + 1
    if char in "aeiouAEIOU":
        vow = vow + 1
    else:
        cons = cons + 1
print("count =",count)
print("vowels =",vow)
print("consonants =",cons)
```
